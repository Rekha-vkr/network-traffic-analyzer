```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("../data/Tuesday-WorkingHours.pcap_ISCX.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Handle missing values
df["Flow Bytes/s"] = df["Flow Bytes/s"].fillna(0)

# Remove duplicate rows
df = df.drop_duplicates()

# Remove invalid negative Flow Duration values
df = df[df["Flow Duration"] >= 0].copy()

print("Cleaned dataset shape:", df.shape)
print("Minimum Flow Duration:", df["Flow Duration"].min())
print("Negative Flow Duration records:", (df["Flow Duration"] < 0).sum())
```

```python
# Flow Duration statistics by traffic type

duration_by_label = df.groupby("Label")["Flow Duration"].describe()

print("Flow Duration statistics by traffic type:")
print(duration_by_label)
```

```python
# Average Flow Duration by traffic type
import matplotlib.pyplot as plt
average_duration = df.groupby("Label")["Flow Duration"].mean()

print("Average Flow Duration by traffic type:")
print(average_duration)

plt.figure(figsize=(7, 5))
average_duration.plot(kind="bar")

plt.title("Average Flow Duration by Traffic Type")
plt.xlabel("Traffic Type")
plt.ylabel("Average Flow Duration")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("../results/graphs/average_flow_duration.png", dpi=300)
plt.show()
```

```python
# Packet statistics by traffic type

packet_stats = df.groupby("Label")[
    ["Total Fwd Packets", "Total Backward Packets"]
].mean()

print("Average packet counts by traffic type:")
print(packet_stats)
```

```python
# Average packet counts by traffic type

packet_stats.plot(kind="bar", figsize=(8, 5))

plt.title("Average Packet Counts by Traffic Type")
plt.xlabel("Traffic Type")
plt.ylabel("Average Number of Packets")
plt.xticks(rotation=0)
plt.legend(["Forward Packets", "Backward Packets"])

plt.tight_layout()
plt.savefig("../results/graphs/average_packet_counts.png", dpi=300)
plt.show()
```

```python
# Average Flow Bytes/s by traffic type

average_flow_bytes = df.groupby("Label")["Flow Bytes/s"].mean()

print("Average Flow Bytes/s by traffic type:")
print(average_flow_bytes)
```

```python
# Check infinite values in Flow Bytes/s

infinite_flow_bytes = np.isinf(df["Flow Bytes/s"]).sum()

print("Infinite Flow Bytes/s values:", infinite_flow_bytes)
```

```python
# Handle infinite values in Flow Bytes/s

df["Flow Bytes/s"] = df["Flow Bytes/s"].replace([np.inf, -np.inf], np.nan)

# Calculate median using valid values only
flow_bytes_median = df["Flow Bytes/s"].median()

# Fill missing values with the median
df["Flow Bytes/s"] = df["Flow Bytes/s"].fillna(flow_bytes_median)

print("Flow Bytes/s median:", flow_bytes_median)
print("Missing values:", df["Flow Bytes/s"].isna().sum())
print("Infinite values:", np.isinf(df["Flow Bytes/s"]).sum())
```

```python
average_flow_bytes = df.groupby("Label")["Flow Bytes/s"].mean()

print("Average Flow Bytes/s by traffic type:")
print(average_flow_bytes)
```

```python
# Average Flow Bytes/s by traffic type

average_flow_bytes.plot(kind="bar", figsize=(8, 5))

plt.title("Average Flow Bytes/s by Traffic Type")
plt.xlabel("Traffic Type")
plt.ylabel("Average Flow Bytes/s")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("../results/graphs/average_flow_bytes.png", dpi=300)
plt.show()
```

```python
# Average Flow Packets/s by traffic type

average_flow_packets = df.groupby("Label")["Flow Packets/s"].mean()

print("Average Flow Packets/s by traffic type:")
print(average_flow_packets)
```

```python
# Clean infinite Flow Packets/s values

df["Flow Packets/s"] = df["Flow Packets/s"].replace([np.inf, -np.inf], np.nan)

flow_packets_median = df["Flow Packets/s"].median()

df["Flow Packets/s"] = df["Flow Packets/s"].fillna(flow_packets_median)

print("Flow Packets/s median:", flow_packets_median)
print("Missing values:", df["Flow Packets/s"].isna().sum())
print("Infinite values:", np.isinf(df["Flow Packets/s"]).sum())
```

```python
# Average Flow Packets/s by traffic type

average_flow_packets = df.groupby("Label")["Flow Packets/s"].mean()

print("Average Flow Packets/s by traffic type:")
print(average_flow_packets)
```

```python
# Average Flow Packets/s by traffic type

average_flow_packets.plot(kind="bar", figsize=(8, 5))

plt.title("Average Flow Packets/s by Traffic Type")
plt.xlabel("Traffic Type")
plt.ylabel("Average Flow Packets/s")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("../results/graphs/average_flow_packets.png", dpi=300)
plt.show()
```

```python
# Final data quality check

numeric_columns = df.select_dtypes(include=np.number).columns

missing_values = df[numeric_columns].isna().sum().sum()
infinite_values = np.isinf(df[numeric_columns]).sum().sum()

print("Total missing numerical values:", missing_values)
print("Total infinite numerical values:", infinite_values)
```

```python
# Select important network traffic features

correlation_features = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Total Length of Fwd Packets",
    "Total Length of Bwd Packets",
    "Flow Bytes/s",
    "Flow Packets/s"
]

correlation_matrix = df[correlation_features].corr()

print("Correlation matrix:")
print(correlation_matrix.round(2))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 7))

plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_features)),
    correlation_features,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation_features)),
    correlation_features
)

# Add correlation values inside the cells
for i in range(len(correlation_features)):
    for j in range(len(correlation_features)):
        plt.text(
            j,
            i,
            f"{correlation_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Heatmap of Selected Network Features")
plt.tight_layout()

plt.savefig(
    "../results/graphs/correlation_heatmap.png",
    dpi=300
)

plt.show()
```
