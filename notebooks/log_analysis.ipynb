```python
# Separate input features and target label
X = df.drop("Label", axis=1)
y = df["Label"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget classes:")
print(y.value_counts())
```

```python
import sklearn
print(sklearn.__version__)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
import numpy as np

print("Infinite values in X_train:", np.isinf(X_train).sum().sum())
print("Infinite values in X_test:", np.isinf(X_test).sum().sum())
```

```python
inf_columns = X_train.columns[np.isinf(X_train).any()].tolist()

print("Columns containing infinite values:")
print(inf_columns)
```

```python
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)

print("Infinite values in X_train:", np.isinf(X_train).sum().sum())
print("Infinite values in X_test:", np.isinf(X_test).sum().sum())
```

```python
missing_train = X_train.isnull().sum()

print("Columns with missing values in training data:")
print(missing_train[missing_train > 0])
```

```python
missing_test = X_test.isnull().sum()

print("Columns with missing values in testing data:")
print(missing_test[missing_test > 0])
```

```python
# Calculate median values from training data
flow_bytes_median = X_train["Flow Bytes/s"].median()
flow_packets_median = X_train["Flow Packets/s"].median()

print("Flow Bytes/s median:", flow_bytes_median)
print("Flow Packets/s median:", flow_packets_median)
```

```python
# Fill missing values using medians calculated from training data

X_train["Flow Bytes/s"] = X_train["Flow Bytes/s"].fillna(flow_bytes_median)
X_test["Flow Bytes/s"] = X_test["Flow Bytes/s"].fillna(flow_bytes_median)

X_train["Flow Packets/s"] = X_train["Flow Packets/s"].fillna(flow_packets_median)
X_test["Flow Packets/s"] = X_test["Flow Packets/s"].fillna(flow_packets_median)
```

```python
print("Missing values in X_train:", X_train.isnull().sum().sum())
print("Missing values in X_test:", X_test.isnull().sum().sum())

print("Infinite values in X_train:", np.isinf(X_train).sum().sum())
print("Infinite values in X_test:", np.isinf(X_test).sum().sum())
```
